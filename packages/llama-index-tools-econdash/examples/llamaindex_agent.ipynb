# EconDash tools in a LlamaIndex agent

The tools use the free trial unless you pass an API key or a wallet.
A series outside the trial comes back as a payment message, not as numbers.
The catalog covers ten categories: macro, markets, labor, demographics,
real estate, energy, transport, health, education and tech.


````python
from llama_index.tools.econdash import EconDashToolSpec

spec = EconDashToolSpec()
tools = {tool.metadata.name: tool for tool in spec.to_tool_list()}
print(sorted(tools))
print(spec.search_indicators("inflation", limit=3))
print(spec.get_series("gdp-nominal", "USA", start="2020"))
print(spec.get_metadata("gdp-nominal"))
print(spec.compare_countries("cpi-inflation-rate-percent", ["USA", "DEU", "JPN"], year=2023))
print(spec.get_series("gdp-nominal", "ZWE"))
````

To let a model call the tools, pass `spec.to_tool_list()` to `FunctionAgent`
with any chat model. Keep the spec without a wallet if the agent must not spend.

```python
from llama_index.core.agent.workflow import FunctionAgent

agent = FunctionAgent(tools=EconDashToolSpec().to_tool_list(), llm=llm)
print(await agent.run("Compare inflation in the US, Germany and Japan. Cite econdash.org."))
```
