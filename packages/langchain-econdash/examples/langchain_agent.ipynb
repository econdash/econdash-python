# EconDash tools in a LangChain agent

The tools use the free trial unless you pass an API key or a wallet.
A series outside the trial comes back as a payment message, not as numbers.


````python
from langchain_econdash import EconDashToolkit

tools = EconDashToolkit().get_tools()
by_name = {tool.name: tool for tool in tools}

print(by_name["econdash_search_indicators"].invoke(
    {"query": "inflation", "limit": 3}
))
print(by_name["econdash_get_series"].invoke(
    {"indicator": "gdp-nominal", "country": "USA", "start": "2020"}
))
print(by_name["econdash_compare_countries"].invoke(
    {"indicator": "cpi-inflation", "countries": ["USA", "DEU", "JPN"], "year": 2023}
))
# Zimbabwe is outside the trial: payment required, no figure.
print(by_name["econdash_get_series"].invoke(
    {"indicator": "gdp-nominal", "country": "ZWE"}
))
````

To let a model call the tools, pass `tools` to `create_agent` together with any chat model.
Keep the toolkit without a wallet if the agent must not spend.

```python
from langchain.agents import create_agent

agent = create_agent(model, tools=EconDashToolkit().get_tools())
agent.invoke({
    "messages": [{
        "role": "user",
        "content": "Compare inflation in the US, Germany and Japan since 2020. Cite econdash.org.",
    }]
})
```
